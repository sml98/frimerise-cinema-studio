```python
#@title 3. Produzir Documentário Cinema v8 (salva automaticamente em Meu Drive/FrimeRise/)
%cd /content/frimerise-cinema-studio

TOPIC = "A Verdadeira Hist\u00f3ria Oculta da Biblioteca de Alexandria" #@param {type:"string"}
THUMB_3WORDS = "O INC\u00caNDIO OCULTO" #@param {type:"string"}
ASPECT = "16:9" #@param ["16:9", "9:16"]
MINUTES = 17 #@param {type:"number"}
LANG = "pt_br" #@param ["pt_br", "en_us", "es"]
SHORTS_CASCADE = False #@param {type:"boolean"}

cmd = [
    "python", "batch_produce.py",
    "--topic", TOPIC,
    "--engine", "netflix_doc",
    "--aspect", ASPECT,
    "--minutes", str(MINUTES),
    "--lang", LANG,
    "--thumb-3words", THUMB_3WORDS,
]
if SHORTS_CASCADE:
    cmd.append("--shorts-cascade")

import subprocess, sys
rc = subprocess.call(cmd)
if rc != 0:
    sys.exit(rc)
print('🎉 Vídeo finalizado e arquivado em Meu Drive/FrimeRise/!')
```
